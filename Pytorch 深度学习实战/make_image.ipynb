```python
import numpy as np
from PIL import Image # A library to handle images

# 1. Create a 3D array (height, width, channels) with all zeros (black)
# The data type 'uint8' is an 8-bit integer, perfect for colors from 0-255.
height = 60
width = 120
channels = 3 # For RGB (Red, Green, Blue)

# This creates our grid of numbers
black_image_data = np.zeros((height, width, channels), dtype=np.uint8)

# Let's make the image red instead by changing the numbers!
# We'll set the Red channel (the 0th channel) to its max value, 255.
black_image_data[:, :, 0] = 255 

# 2. Convert the NumPy array into an actual image file
img = Image.fromarray(black_image_data, 'RGB')
img.save('my_image.png')

print("An image named 'my_image.png' has been created!")
```
