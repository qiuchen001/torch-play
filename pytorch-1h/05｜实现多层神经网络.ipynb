```python
import torch
import torch.nn.functional as F

class NeuralNetwork(torch.nn.Module):
    def __init__(self):
        super(NeuralNetwork, self).__init__()
        self.linear1 = torch.nn.Linear(2, 2)
        self.linear2 = torch.nn.Linear(2, 1)
        
    def forward(self, x):
        x = self.linear1(x)
        x = F.relu(x)
        
        
        x = self.linear2(x)
        x = torch.sigmoid(x)
        return x
    
model = NeuralNetwork()
print(model)
```

```python
y = torch.tensor([[1.0]], dtype=torch.float32)
x1 = torch.tensor([[1.0, 2.0]], dtype=torch.float32)

y.shape

```

```python
logists = model(x1)
loss = F.binary_cross_entropy(logists, y)
print(loss)
```

```python
from torch.autograd import grad
grad(loss, model.parameters(), retain_graph= True)
```

```python
model.parameters()
```

```python
loss.backward()
```

```python
class NeuralNetwork1(torch.nn.Module):
    def __init__(self):
        super().__init__()
        self.layers = torch.nn.Sequential(
            torch.nn.Linear(2, 2),
            torch.nn.ReLU(),
            torch.nn.Linear(2, 1),
            torch.nn.Sigmoid()
        )
        
    def forward(self, x):
        return self.layers(x)
        
```
